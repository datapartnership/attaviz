# Altair Theme — Gallery

```python
import altair as alt
import pandas as pd
from vega_datasets import data

import altairtheme

altairtheme.enable()
```

## 1. Bar Chart — Country Comparison

Bar charts are recommended for **ranking or comparing values** across categories.

```python
source = pd.DataFrame(
    {
        "country": [
            "India",
            "China",
            "Nigeria",
            "Indonesia",
            "Brazil",
            "Pakistan",
            "Bangladesh",
            "Ethiopia",
        ],
        "population": [1428, 1425, 224, 277, 216, 230, 173, 126],
    }
)

alt.Chart(
    source,
    title=alt.Title(
        "Most Populous Countries", subtitle="Population in millions, 2023 estimates"
    ),
).mark_bar().encode(
    x=alt.X("population:Q", title="Population (millions)"),
    y=alt.Y("country:N", sort="-x", title=None),
    color=alt.value(altairtheme.CATEGORICAL[0]),
).properties(width=500, height=300)
```

## 2. Line Chart — Temporal Trends

Line charts show temporal trends. The theme applies a 4px stroke with round caps.

```python
source = pd.DataFrame(
    {
        "year": list(range(2015, 2024)) * 3,
        "region": ["East Asia & Pacific"] * 9
        + ["South Asia"] * 9
        + ["Sub-Saharan Africa"] * 9,
        "gdp_growth": [
            6.1,
            6.0,
            6.1,
            5.9,
            5.5,
            1.2,
            7.5,
            3.5,
            5.1,
            7.2,
            7.6,
            6.4,
            6.3,
            3.6,
            -4.5,
            7.9,
            5.8,
            5.5,
            3.0,
            1.3,
            2.6,
            3.2,
            2.7,
            -1.9,
            4.4,
            3.7,
            3.5,
        ],
    }
)

alt.Chart(
    source,
    title=alt.Title(
        "GDP Growth by Region", subtitle="Annual GDP growth rate (%), 2015–2023"
    ),
).mark_line().encode(
    x=alt.X("year:O", title=None),
    y=alt.Y("gdp_growth:Q", title="GDP Growth (%)"),
    color=alt.Color("region:N", title="Region"),
).properties(width=550, height=350)
```

## 3. Scatter Plot

Scatter plots display relationships between two indicators. Points get a 1px white
outline for clarity, as specified in the style guide.

```python
source = data.cars()

alt.Chart(
    source,
    title=alt.Title(
        "Horsepower vs. Fuel Efficiency",
        subtitle="Relationship between engine power and mileage by origin",
    ),
).mark_circle(size=80).encode(
    x=alt.X("Horsepower:Q"),
    y=alt.Y("Miles_per_Gallon:Q", title="Miles per Gallon"),
    color=alt.Color("Origin:N", title="Origin"),
    tooltip=["Name:N", "Horsepower:Q", "Miles_per_Gallon:Q"],
).properties(width=550, height=380)
```

## 5. Area Chart — Stacked Composition

Stacked areas show part-to-whole relationships over time.

```python
source = data.iowa_electricity()

alt.Chart(
    source,
    title=alt.Title(
        "Iowa Electricity Generation", subtitle="Net generation by source (GWh)"
    ),
).mark_area().encode(
    x=alt.X("year:T", title=None),
    y=alt.Y("net_generation:Q", title="Net Generation (GWh)"),
    color=alt.Color("source:N", title="Source"),
).properties(width=550, height=350)
```

## 6. Heatmap — Sequential Color Scale

The WBG blue sequential palette (`SEQ_BLUE`) is the default for ordered
quantitative color scales.

```python
source = data.seattle_weather()

alt.Chart(
    source,
    title=alt.Title(
        "Seattle Monthly Max Temperatures",
        subtitle="Average daily maximum temperature (°F) by month",
    ),
).mark_rect().encode(
    x=alt.X("month(date):O", title="Month"),
    y=alt.Y("year(date):O", title="Year", axis=alt.Axis(grid=False)),
    color=alt.Color("mean(temp_max):Q", title="Avg Max Temp (°F)"),
).properties(width=500, height=250)
```

## 7. Histogram

Showing distributions with automatic binning.

```python
source = data.movies()

alt.Chart(
    source,
    title=alt.Title(
        "Distribution of IMDB Ratings",
        subtitle="Histogram of movie ratings from the IMDB database",
    ),
).mark_bar().encode(
    x=alt.X("IMDB_Rating:Q", bin=alt.Bin(maxbins=30), title="IMDB Rating"),
    y=alt.Y("count()", title="Number of Movies"),
    color=alt.value(altairtheme.CATEGORICAL[0]),
).properties(width=550, height=300)
```

## 8. Using Custom Sequential Palettes

You can swap in any WBG sequential palette by passing the color list to
`alt.Scale(range=...)`.

```python
source = data.cars()

alt.Chart(
    source,
    title=alt.Title(
        "Acceleration by Horsepower and Cylinders",
        subtitle="Green sequential palette for favorable higher values",
    ),
).mark_circle(size=80).encode(
    x=alt.X("Horsepower:Q"),
    y=alt.Y("Acceleration:Q"),
    color=alt.Color(
        "Acceleration:Q",
        scale=alt.Scale(range=altairtheme.SEQ_GREEN),
        title="Acceleration",
    ),
).properties(width=550, height=350)
```

## 9. Diverging Color Scale

Use diverging palettes for data with a meaningful midpoint. The default WBG
diverging palette (`DIV_DEFAULT`) implies good (blue) vs. bad (red).

```python
import numpy as np

np.random.seed(42)
countries = ["BRA", "CHN", "IND", "IDN", "MEX", "NGA", "ZAF", "TUR", "VNM", "EGY"]
source = pd.DataFrame(
    {
        "country": countries,
        "change": np.random.uniform(-8, 8, len(countries)).round(1),
    }
)

alt.Chart(
    source,
    title=alt.Title(
        "GDP Growth Change",
        subtitle="Year-over-year change in GDP growth rate (percentage points)",
    ),
).mark_bar().encode(
    x=alt.X("change:Q", title="Change (pp)"),
    y=alt.Y("country:N", sort="-x", title=None),
    color=alt.Color(
        "change:Q",
        scale=alt.Scale(
            range=altairtheme.DIV_DEFAULT,
            domainMid=0,
        ),
        title="Change",
    ),
).properties(width=500, height=300)
```

## 10. Faceted Small Multiples

The theme styles facet headers consistently with the WBG typography.

```python
source = data.iris()

alt.Chart(
    source,
).mark_circle(size=60).encode(
    x=alt.X("sepalLength:Q", title="Sepal Length"),
    y=alt.Y("petalLength:Q", title="Petal Length"),
    color=alt.Color("species:N", title="Species"),
).properties(
    width=200,
    height=200,
).facet(
    column=alt.Column("species:N", title=None),
).properties(
    title=alt.Title(
        "Iris Measurements by Species", subtitle="Sepal vs. petal dimensions"
    )
)
```

## 11. Title, Subtitle, and Caption

Altair natively supports **title** and **subtitle** via `alt.Title`. For a
**caption** (source note at the bottom), concatenate a text mark below the chart.
This pattern is useful for adding data sources, methodology notes, or disclaimers.

```python
source = pd.DataFrame(
    {
        "country": [
            "India",
            "China",
            "Nigeria",
            "Indonesia",
            "Brazil",
            "Pakistan",
            "Bangladesh",
            "Ethiopia",
        ],
        "population": [1428, 1425, 224, 277, 216, 230, 173, 126],
    }
)

chart = (
    alt.Chart(
        source,
        title=alt.Title(
            "Most Populous Countries",
            subtitle="Population in millions, 2023 estimates",
        ),
    )
    .mark_bar()
    .encode(
        x=alt.X("population:Q", title="Population (millions)"),
        y=alt.Y("country:N", sort="-x", title=None),
        color=alt.value(altairtheme.CATEGORICAL[0]),
    )
    .properties(width=500, height=300)
)

caption = (
    alt.Chart({"values": [{}]})
    .mark_text(
        align="right",
        font="Open Sans, sans-serif",
        fontSize=12,
        color=altairtheme.TEXT_SUBTLE,
    )
    .encode(text=alt.value("Source: United Nations, World Population Prospects 2023"))
    .properties(width=500)
)

alt.vconcat(chart, caption, spacing=21)
```

## 12. Color Palette Reference

Visual reference of all available palettes.

```python
def swatch(name, colors):
    """Create a horizontal color swatch."""
    df = pd.DataFrame(
        {
            "color": colors,
            "x": range(len(colors)),
        }
    )
    return (
        alt.Chart(df, title=name)
        .mark_rect(stroke=None, cornerRadius=4)
        .encode(
            x=alt.X("x:O", axis=None),
            color=alt.Color(
                "color:N", scale=alt.Scale(domain=colors, range=colors), legend=None
            ),
        )
        .properties(width=300, height=40)
    )


swatches = alt.vconcat(
    swatch("Categorical", altairtheme.CATEGORICAL),
    swatch("Sequential: Blue", altairtheme.SEQ_BLUE),
    swatch("Sequential: Green", altairtheme.SEQ_GREEN),
    swatch("Sequential: Purple", altairtheme.SEQ_PURPLE),
    swatch("Sequential: Yellow", altairtheme.SEQ_YELLOW),
    swatch("Sequential: Red", altairtheme.SEQ_RED),
    swatch("Sequential: Bad to Good", altairtheme.SEQ_BAD_TO_GOOD),
    swatch("Sequential: Good to Bad", altairtheme.SEQ_GOOD_TO_BAD),
    swatch("Diverging: Default", altairtheme.DIV_DEFAULT),
    swatch("Diverging: Neutral", altairtheme.DIV_NEUTRAL),
    swatch("Diverging: Alternative", altairtheme.DIV_ALTERNATIVE),
    swatch("Income Groups", altairtheme.INCOME_LIST),
    swatch("Age Groups", altairtheme.AGE),
    title="WBG Color Palettes",
)
swatches
```

## 13. Responsive Sizing

The WBG theme supports responsive typography and spacing based on chart width.
Three size categories are available: **small** (<400px), **medium** (400-700px),
and **large** (>700px). Each category has appropriately scaled font sizes and spacing.

```python
# View the typography and spacing scales
print("Typography (font sizes by category):")
for size, fonts in altairtheme.TYPOGRAPHY.items():
    print(f"  {size}: S={fonts['S']}px, M={fonts['M']}px, L={fonts['L']}px")

print("\nSpacing (padding values by category):")
for size, spaces in altairtheme.SPACING.items():
    print(f"  {size}: {spaces}")

print("\nDefault dimensions:")
for size, dims in altairtheme.DEFAULT_DIMENSIONS.items():
    print(f"  {size}: {dims[0]}x{dims[1]}px")
```

### 13.1 Size-Specific Global Themes

Use `enable(size='...')` to switch the global theme to a specific size category.
This affects all charts created afterwards.

```python
# Sample data for size comparison
size_demo_data = pd.DataFrame(
    {
        "category": ["A", "B", "C", "D", "E"],
        "value": [28, 55, 43, 91, 81],
    }
)


def make_demo_chart(title_text):
    """Create a simple bar chart for size demonstration."""
    return (
        alt.Chart(size_demo_data, title=alt.Title(title_text, subtitle="Subtitle text"))
        .mark_bar()
        .encode(
            x=alt.X("value:Q", title="Value"),
            y=alt.Y("category:N", title=None),
            color=alt.value(altairtheme.CATEGORICAL[0]),
        )
    )
```

```python
# Small theme (< 400px width) — compact typography for mobile/thumbnails
altairtheme.enable(size="small")

make_demo_chart("Small Theme (350x233)").properties(width=350, height=233)
```

```python
# Medium theme (400-700px width) — standard typography for most use cases
altairtheme.enable(size="medium")

make_demo_chart("Medium Theme (600x400)").properties(width=600, height=400)
```

```python
# Large theme (> 700px width) — larger typography for presentations/reports
altairtheme.enable(size="large")

make_demo_chart("Large Theme (800x533)").properties(width=800, height=533)
```

### 13.2 Per-Chart Sizing with `configure_size()`

Use `configure_size()` to apply size-appropriate typography to individual charts
without changing the global theme. This is useful for dashboards with charts of
varying sizes.

```python
# Reset to medium theme for the rest of the notebook
altairtheme.enable(size="medium")

# Create a base chart
base_chart = (
    alt.Chart(
        size_demo_data,
        title=alt.Title("Per-Chart Sizing", subtitle="Using configure_size()"),
    )
    .mark_bar()
    .encode(
        x=alt.X("value:Q", title="Value"),
        y=alt.Y("category:N", title=None),
        color=alt.value(altairtheme.CATEGORICAL[1]),
    )
)

# Apply small-size typography to a 300px wide chart
altairtheme.configure_size(base_chart, width=300, aspect_ratio="4:3")
```

### 13.3 Aspect Ratio Utilities

Use `calculate_dimensions()` to compute height from width and a named aspect ratio.
Available ratios: `'16:9'`, `'4:3'`, `'3:2'`, `'1:1'`, `'2:1'`, `'square'`, `'widescreen'`.

```python
# Calculate dimensions for common aspect ratios
print("Calculating dimensions from width=800:")
for ratio in ["16:9", "4:3", "3:2", "1:1", "2:1"]:
    w, h = altairtheme.calculate_dimensions(800, ratio)
    print(f"  {ratio:5s} → {w}x{h}")

# Available aspect ratios
print("\nAvailable aspect ratios:")
for name, value in altairtheme.ASPECT_RATIOS.items():
    print(f"  {name}: {value:.3f}")
```

### 13.4 Custom Theme Registration

Use `wbg_theme()` with parameters to create custom theme configurations:

```python
# Register a custom widescreen theme for presentations
alt.themes.register(
    "wbg-widescreen",
    lambda: altairtheme.wbg_theme(width=1200, aspect_ratio="16:9"),
)

# Enable and use the custom theme
alt.themes.enable("wbg-widescreen")

# Inspect the theme configuration
theme = altairtheme.wbg_theme(width=1200, aspect_ratio="16:9")
print(
    f"Custom theme dimensions: {theme['config']['view']['continuousWidth']}x{theme['config']['view']['continuousHeight']}"
)
print(
    f"Typography (large): title={theme['config']['title']['fontSize']}px, labels={theme['config']['axis']['labelFontSize']}px"
)

# Reset to default medium theme
altairtheme.enable()
```
